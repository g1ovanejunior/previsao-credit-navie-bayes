```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import confusion_matrix, accuracy_score
from yellowbrick.classifier import ConfusionMatrix
from scipy.stats import chi2_contingency
from sklearn.feature_selection import mutual_info_classif
from sklearn.preprocessing import OrdinalEncoder
```

```python
credito = pd.read_csv("dados/Credit.csv")
credito.head()
```

```python
#Separar os tipos de coluna entre: numéricos, categóricos ordinais e categóricos nominais

# Separar a variável que representa o resultado (y) das variáveis que são só características (X)
X = credito.drop(columns=['class'])
y = credito['class']

# 1. armazenando apenas as colunas numéricas, as que usam números inteiros e números float
colunas_numericas = X.select_dtypes(include=['int64', 'float64']).columns.tolist()

# 2. armazenando as colunas categóricas, utilizando essas classes que indicam textos.
colunas_categoricas = X.select_dtypes(include=['string','object', 'category']).columns.tolist()

print("Numéricas:", colunas_numericas)
print("Categóricas:", colunas_categoricas)
```

```python
#Separar as categóricas entre Ordinais e Nominais
colunas_ordinais = ['checking_status', 'credit_history' , 'savings_status', 'employment', 'property_magnitude' ]
colunas_nominais = [col for col in colunas_categoricas if col not in colunas_ordinais]

print("Nominais:", colunas_nominais)
print("Ordinais:", colunas_ordinais)
print("Numéricas:", colunas_numericas)
```

```python
#Analisando se há a falta de informação em alguma parte do banco de dados.
credito.isnull().sum()
#no caso, não há
```

```python
#Calcular a proporção entre os resultados da classe (y).Isso é feito para evitar uma desregulação no resultado da IA preditiva e buscar uma porcentagem
#correspondente ao real resultado esperado.
proporcao = credito['class'].value_counts(normalize=True) * 100
print(proporcao)
#A accuracy da IA sem análise acerta 71% das vezes, mas como 70% da classe é "good", então ela só precisa afirmar que todas são "good", e terá 70% de acerto.
```

```python
#Loop para testar nível de significância entre cada coluna e a classe.É usado 5% como métrica para decidir se impacta ou não no resultado.
# Se o p_value for >0,005 significa que não impacta na classe.

# Listas para armazenar o resultado da análise
colunas_descartar = []

# 2. Loop passando por todas as colunas categóricas
for coluna in colunas_categoricas:
    # Cria a tabela de contingência
    tabela_cruzada = pd.crosstab(credito[coluna], credito['class'])
    
    # Executa o teste do Qui-Quadrado
    stat, p_valor, dof, esperados = chi2_contingency(tabela_cruzada)
    
    # Verifica se o p-valor é MAIOR que 0.05 (5%)
    if p_valor > 0.05:
        colunas_descartar.append(coluna)
        print(f" Coluna '{coluna}': p-value = {p_valor:.4f} (MAIOR que 5% ")
```

```python
#Excluir as colunas que possuem pouco nível de relevância pro resultado, ou que são resultados do acaso.
credito_filtrado1 = credito.drop(columns=colunas_descartar)
credito_filtrado1.head()
```

```python
#Observa-se que foi diminuído 2 colunas.
credito_filtrado1.shape
```

```python
#Utilizar o parâmetro Informação Mútua pra decidir se ainda há alguma coluna que não apresente relevância pra classe.
 
#Separar X (atributos) e y (classe) a partir do seu dataset filtrado
X = credito_filtrado1.drop(columns=['class'])
y = credito_filtrado1['class']

#Criar uma cópia e codificar as categóricas temporariamente
X_encoded = X.copy()
colunas_cat_filtradas1 = X.select_dtypes(include=['string', 'object', 'category']).columns

encoder = OrdinalEncoder()
X_encoded[colunas_cat_filtradas1] = encoder.fit_transform(X[colunas_cat_filtradas1])

#Mapear os índices das variáveis discretas/categóricas para o algoritmo
indices_categoricas = [X.columns.get_loc(col) for col in colunas_cat_filtradas1]

#Calcular a Informação Mútua
mi_scores = mutual_info_classif(
    X_encoded, 
    y, 
    discrete_features=indices_categoricas, 
    random_state=42
)

#Criar uma Série organizada do maior para o menor valor
mi_series = pd.Series(mi_scores, index=X.columns).sort_values(ascending=False)

#Exibir os valores calculados
print(mi_series)
```

```python
plt.figure(figsize=(10, 6))
mi_series.sort_values().plot(kind='barh', color='darkcyan')
plt.title("Importância das Variáveis (Informação Mútua)")
plt.xlabel("Score de Informação Mútua (Quanto maior, mais importante)")
plt.ylabel("Atributos")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```
